```python
import pandas as pd
nav_history = pd.read_csv("../data/raw/02_nav_history.csv")
nav_history.info()
```

```python
nav_history.describe(include="all")
```

```python
nav_history.isnull().sum()
```

```python
nav_history.dtypes
```

```python
nav_history.duplicated().sum()
```

```python
# Date is in string format, we need to convert it to datetime format
nav_history["date"] = pd.to_datetime(nav_history["date"])
print(nav_history["date"].dtype)
```

```python
nav_history.sort_values(by=["amfi_code", "date"])
nav_history.head()
```

```python
if nav_history["nav"].isnull().sum() > 0:
    nav_history["nav"] = nav_history.groupby("amfi_code")["nav"].ffill()
print("Missing NAV values after forward fill:", nav_history["nav"].isnull().sum())
```

```python
#validate NAV > 0
print((nav_history["nav"] <= 0).sum())
```

```python
nav_history.to_csv("../data/processed/02_nav_history_cleaned.csv", index=False)
```

## 7th Dataset

```python
scheme_performance = pd.read_csv("../data/raw/07_scheme_performance.csv")
scheme_performance.info()
```

```python
scheme_performance.isnull().sum()
```

```python
scheme_performance.duplicated().sum()
```

```python
# Validate return values are numeric
scheme_performance[["return_1yr_pct","return_3yr_pct","return_5yr_pct","benchmark_3yr_pct"]].dtypes
# Return values are already in numeric format, no further cleaning is required.
# We can also check for any invalid return values (non-numeric) and handle them if necessary. However, since the return values are already in numeric format, we can skip this step.
#  return_cols = [
#     "return_1yr_pct",
#     "return_3yr_pct",
#     "return_5yr_pct",
#     "benchmark_3yr_pct"
# ]

# for col in return_cols:
#     scheme_performance[col] = pd.to_numeric(scheme_performance[col], errors="coerce")

# invalid = scheme_performance[scheme_performance[return_cols].isna().any(axis=1)]
```

```python
#flag anomalies in return values
anomalies = scheme_performance[(scheme_performance["return_1yr_pct"] < -100) | (scheme_performance["return_1yr_pct"] > 100) |
                               (scheme_performance["return_3yr_pct"] < -100) | (scheme_performance["return_3yr_pct"] > 100) |
                               (scheme_performance["return_5yr_pct"] < -100) | (scheme_performance["return_5yr_pct"] > 100) |
                               (scheme_performance["benchmark_3yr_pct"] < -100) | (scheme_performance["benchmark_3yr_pct"] > 100)]

print("Number of anomalies in return values:",anomalies)
```

```python
# check expense_ratio range (0.1% – 2.5%).
invalid_expense_ratio = scheme_performance[(scheme_performance["expense_ratio_pct"] < 0.1) | (scheme_performance["expense_ratio_pct"] > 2.5)]
print("Number of invalid expense ratios:",invalid_expense_ratio)
print("Number of invalid expense ratios:",invalid_expense_ratio.shape[0])
print("Number of invalid expense ratios:",len(invalid_expense_ratio))
```

```python
scheme_performance.to_csv("../data/processed/07_scheme_performance_cleaned.csv", index=False)
```
