```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
engine = create_engine("sqlite:///../database/bluestock_mf.db")
```

```python
fund_master = pd.read_csv("../data/processed/01_fund_master_cleaned.csv")
```

```python
fund_master.to_sql("dim_fund", con=engine, if_exists="replace", index=False)
```

```python
csv_rows = len(fund_master)

db_rows = pd.read_sql(
    "SELECT COUNT(*) AS total FROM dim_fund",
    engine
)

print("CSV Rows:", csv_rows)
print("Database Rows:", db_rows["total"][0])
```
